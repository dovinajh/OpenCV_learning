```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
def cv_show(img,name):
    cv2.imshow(name,img)
    cv2.waitKey()
    cv2.destroyAllWindows()
```

```python
img=cv2.imread(r'C:\Users\Dovina\Desktop\lenaNoise.png',cv2.IMREAD_GRAYSCALE)

v1=cv2.Canny(img,80,150)
v2=cv2.Canny(img,50,100)

res=np.hstack((v1,v2))
cv_show(res,'res')
```
