```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
def cv_show(img,name):
    cv2.imshow(name,img)
    cv2.waitKey()
    cv2.destroyAllWindows()
```

```python
img=cv2.imread(r'C:\Users\Dovina\Desktop\contours.jpg')
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
ret,thresh=cv2.threshold(gray,127,255,cv2.THRESH_BINARY)
cv_show(thresh,'thresh')
```

```python
contours,hierarchy=cv2.findContours(thresh,cv2.RETR_TREE,cv2.CHAIN_APPROX_NONE)
```

# 绘制轮廓

```python
cv_show(img,'img')
```

```python
#传入绘制图像，轮廓，轮廓索引，颜色模式，线条厚度
#注意需要copy，要不原图会变。。。
draw_img=img.copy()
res=cv2.drawContours(draw_img,contours,-1,(0,0,225),2)
cv_show(res,'res')
```

# 轮廓近似

```python
img = cv2.imread(r'C:\Users\Dovina\Desktop\contours2.png')
down=cv2.pyrDown(img)
cv_show(down,'down')
print(down.shape)
```

```python
down2=cv2.pyrDown(down)
cv_show(down2,'down2')
```

```python
gray = cv2.cvtColor(down2, cv2.COLOR_BGR2GRAY)
ret, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_NONE)
cnt = max(contours, key = lambda x: cv2.contourArea(x))

draw_down2 =down2.copy()
res = cv2.drawContours(draw_down2, [cnt], -1, (0, 0, 255), 2)
cv_show(res,'res')
```

```python
epsilon = 0.15*cv2.arcLength(cnt, True)
approx = cv2.approxPolyDP(cnt, epsilon, True)

draw_down2 = down2.copy()
res = cv2.drawContours(draw_down2, [approx], -1, (0, 0, 255), 2)
cv_show(res,'res')
```

# 边界矩形

```python
img = cv2.imread(r'C:\Users\Dovina\Desktop\contours.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
ret, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_NONE)
cnt = contours[0]

x, y, w, h = cv2.boundingRect(cnt)
img = cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)
cv_show(img,'img')
```

```python
area = cv2.contourArea(cnt)
x, y, w, h = cv2.boundingRect(cnt)
rect_area = w * h
extent = float(area) / rect_area
print ('轮廓面积与边界矩形比', extent)
```

# 外接圆

```python
(x, y), radius = cv2.minEnclosingCircle(cnt)
center = (int(x), int(y))
radius = int(radius)
img = cv2.circle(img, center, radius, (0, 255, 0), 2)
cv_show(img,'img')
```
