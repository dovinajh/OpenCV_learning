```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
def cv_show(img,name):
    cv2.imshow(name,img)
    cv2.waitKey()
    cv2.destroyAllWindows()
```

```python
img=cv2.imread(r'C:\Users\Dovina\Desktop\AM.png')
cv_show(img,'img')
print(img.shape)
```

```python
up=cv2.pyrUp(img)
cv_show(up,'up')
print(up.shape)
```

```python
down=cv2.pyrDown(img)
cv_show(down,'down')
print(down.shape)
```

```python
up2=cv2.pyrUp(up)
cv_show(up2,'up2')
print(up2.shape)
```

```python
up=cv2.pyrUp(img)
up_down=cv2.pyrDown(up)
cv_show(up_down,'up_down')
```

```python
cv_show(np.hstack((img,up_down)),'up_show')
```

```python
up=cv2.pyrUp(img)
up_down=cv2.pyrDown(up)
cv_show(img-up_down,'img-up_down')
```
