```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
img=cv2.imread(r'C:\Users\Dovina\Desktop\pie.jpg',cv2.IMREAD_GRAYSCALE)
cv2.imshow('img',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
def cv_show(img,name):
    cv2.imshow(name,img)
    cv2.waitKey()
    cv2.destroyAllWindows()
```

```python
sobelx=cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)

cv_show(sobelx,'sobelx')
```

```python
sobelx=cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
soblex=cv2.convertScaleAbs(sobelx)
cv_show(sobelx,'sobelx')
```

```python
sobely=cv2.Sobel(img,cv2.CV_64F,0,1,ksize=3)
sobley=cv2.convertScaleAbs(sobely)
cv_show(sobely,'sobely')
```

```python
sobelxy=cv2.addWeighted(sobelx,0.5,sobely,0.5,0)
cv_show(sobelxy,'sobelxy')
```

```python
sobelxy=cv2.Sobel(img,cv2.CV_64F,1,1,ksize=3)
soblexy=cv2.convertScaleAbs(sobelxy)
cv_show(sobelxy,'sobelxy')
```

```python
img = cv2.imread(r'C:\Users\Dovina\Desktop\lenaNoise.png', cv2.IMREAD_GRAYSCALE)
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobelx = cv2.convertScaleAbs(sobelx)
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
sobely = cv2.convertScaleAbs(sobely)
sobelxy = cv2.addWeighted(sobelx, 0.5, sobely, 0.5, 0)
cv_show(sobelxy, 'sobelxy')
```
